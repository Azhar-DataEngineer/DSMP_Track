### `Problem 1` - Print the following pattern. Write a program to use for loop to print the following reverse number pattern.

```bash
5 4 3 2 1
4 3 2 1
3 2 1
2 1
1
```

````python
for i in range(0,5):
  for j in range(5-i,0,-1):
    print(j,sep="\n")
````

````python
for i in range(0,5):
  print(i)
````
