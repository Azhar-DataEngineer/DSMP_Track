### Use Cases of Lists and Tuples in Data Engineering

**Lists (`[]`)** are mutable, meaning their contents can be changed after creation. This makes them ideal for:

*   **Dynamic Data Collections**: When you need to append, remove, or modify elements in a sequence, such as accumulating records during data extraction (ETL) processes, building lists of files to process, or storing temporary results that might change.
*   **Processing Streams**: Handling data streams where new entries arrive continuously and need to be added or modified.
*   **Data Transformation Stages**: Intermediate steps in data pipelines where data structures are actively being manipulated, filtered, or aggregated.

**Tuples (`()`)** are immutable, meaning their contents cannot be changed after creation. This immutability makes them suitable for:

*   **Fixed Records/Schemas**: Representing a fixed-structure record, like a row in a database table or a log entry, where the order and type of elements are known and shouldn't change. For example, a `(timestamp, event_id, user_id)` tuple.
*   **Dictionary Keys**: Because they are hashable (due to immutability), tuples can be used as keys in dictionaries, which is not possible with lists. This is useful for composite keys.
*   **Function Arguments and Return Values**: Functions often return multiple values as a tuple, guaranteeing that the returned values are not accidentally modified by the caller.
*   **Data Integrity**: Ensuring that certain data points remain constant throughout processing, preventing accidental modification, which is critical for data quality and auditing.
*   **Performance and Memory**: As demonstrated earlier, tuples can be slightly more memory-efficient and faster to process in certain scenarios due to their fixed nature.

### Visualizing Use Cases

```python
import matplotlib.pyplot as plt
import pandas as pd

def visualize_list_use_case():
    # Scenario: Collecting log entries dynamically
    log_entries = []
    print("Initial log entries:", log_entries)

    # Simulate adding new log entries over time
    log_entries.append({'timestamp': '2023-01-01 10:00:00', 'level': 'INFO', 'message': 'Process started'})
    log_entries.append({'timestamp': '2023-01-01 10:01:15', 'level': 'DEBUG', 'message': 'Reading config file'})
    log_entries.append({'timestamp': '2023-01-01 10:02:30', 'level': 'ERROR', 'message': 'File not found'})
    print("Log entries after additions:", log_entries)

    # Convert to DataFrame for better visualization
    df_logs = pd.DataFrame(log_entries)

    fig, ax = plt.subplots(figsize=(10, 2))
    ax.axis('off') # Hide axes
    ax.set_title('List Use Case: Dynamic Log Collection')

    # Plotting table
    table = ax.table(cellText=df_logs.values,
                     colLabels=df_logs.columns,
                     loc='center',
                     cellLoc='left')
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1.2, 1.2)

    plt.show()

visualize_list_use_case()
```

```python
def visualize_tuple_use_case():
    # Scenario: Representing sensor readings (fixed schema)
    # Each tuple represents (sensor_id, temperature, humidity, pressure)
    sensor_reading_1 = (101, 25.5, 60.2, 1012.5)
    sensor_reading_2 = (102, 24.9, 61.0, 1011.8)
    sensor_reading_3 = (101, 26.1, 59.8, 1013.0)

    print("Sensor Reading 1 (ID, Temp, Hum, Press):", sensor_reading_1)

    # Attempting to change an element (will raise error if uncommented)
    # sensor_reading_1[1] = 26.0
    # print(sensor_reading_1)

    # Using tuples as dictionary keys for aggregated data
    daily_average_temps = {}
    daily_average_temps[(2023, 1, 1, 101)] = 25.8 # Key: (Year, Month, Day, SensorID)
    daily_average_temps[(2023, 1, 1, 102)] = 24.5
    daily_average_temps[(2023, 1, 2, 101)] = 26.3

    print("\nDaily Average Temperatures (using tuples as keys):")
    for key, value in daily_average_temps.items():
        print(f"Sensor {key[3]} on {key[0]}-{key[1]}-{key[2]}: {value}°C")

    # Visualization of fixed records (simplified)
    data = [
        ['Sensor ID', 'Temperature (°C)', 'Humidity (%)', 'Pressure (hPa)'],
        list(sensor_reading_1),
        list(sensor_reading_2),
        list(sensor_reading_3)
    ]

    fig, ax = plt.subplots(figsize=(10, 3))
    ax.axis('off') # Hide axes
    ax.set_title('Tuple Use Case: Fixed Sensor Records')

    table = ax.table(cellText=data[1:],
                     colLabels=data[0],
                     loc='center',
                     cellLoc='center')
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1.2, 1.2)

    plt.show()

visualize_tuple_use_case()
```

These visualizations illustrate the core difference: lists are flexible containers for evolving data, while tuples are robust, unchangeable records for data that needs to maintain its structure and integrity.

# Sets

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

### Creating Sets

```python
### Creating Sets

# Empty

s = {}
print(s)
print(type(s))


# In python the {} is used for both dictionary and set

s = set() #You need to use set function to create an empty set
print(s)
print(type(s))


#1D set

s1 = {1,2,3}

print(s1)
print(type(s1))


# trying to create 2D set
s2 = {{1,2},{3,4}}

print(s2)
print(type(s2))

#2D set

# s2 = {(1,2), (3,4)}

# print(s2)
# print(type(s2))



```

```python
# homo and heterogenous

#  as python set doesn't allow duplicates it treats 1 and true both as 1 and skips it like it is duplicate

s3 = {1,"hello",4.5, True,(1,2,3)}

print(s3) # they are unordered in case of set


# Using type converstion

s4 = set([1,2,3,4,5,5,])

print(s4)

# Duplicates not allowed

s5 = {1,2,3,4,5,5,5,6,6}

print(s5)

# set can't have mutable items

s6 = {6,"Azhar",4.5,True,(1,2,3),[1,2,3] }

print(s6)


```

```python
s1 = {1,2,3}
s2 = {3,2,1}

print(s1==s2)

# The unordered elements are treated as same if the value remains the same in case of set but in the case of tuple and list it is treated as different


s3 = [1,2,3]
s4 = [3,2,1]

print(s3==s4)

s5 = (1,2,3)
s6 = (3,2,1)

print(s5==s6)
```

### Adding items in the set

```python
s1 = {1,2,3,4}

# add - so adding will add only a single element
s1.add(5)

print(s1)

# update - can add multiple items but need to be as a list

s1.update([6,7,8])

print(s1)
```

###Deleting Items

```python
#del

s1 = {1,2,3}

print(s1)
del s1

# print(s1)

# TypeError: 'set' object doesn't support item deletion
# del s1[0]
# You can delete the entire set but not the elements indiviually

# Discard
s1 = {1,2,3}
s1.discard(3)

print(s1)


# Remove
s1 = {1,2,3}

s1.remove(3)

print(s1)

# s1.remove(7) --> This will give error

# The main difference b/w discard and remove is that when an item is not present in the set and when discard is used it doesn't error out whereas remove does get errored out

# Pop --> Randomluy delete any item

s = {1,2,3,4}

s.pop()

print(s)


# Clear --> Makes it an empty set

s2 = {1,2,3,4}

s2.clear()

print(s2)
```
