### Explanation of XOR Bitwise Operator

The **XOR (exclusive OR)** bitwise operator `^` compares two bits and returns `1` if the bits are different, and `0` if they are the same.

Here's how it works:
- `0 ^ 0 = 0` (bits are the same)
- `0 ^ 1 = 1` (bits are different)
- `1 ^ 0 = 1` (bits are different)
- `1 ^ 1 = 0` (bits are the same)

When applied to numbers, it performs the XOR operation on each corresponding pair of bits in their binary representations.

```python
# Example of XOR bitwise operator

a = 5  # Binary: 0101
b = 3  # Binary: 0011

# Perform XOR operation
result = a ^ b

print(f"a (decimal): {a}, a (binary): {bin(a)}")
print(f"b (decimal): {b}, b (binary): {bin(b)}")
print(f"Result of a ^ b (decimal): {result}")
print(f"Result of a ^ b (binary): {bin(result)}")

# Step-by-step XOR:
#   0101 (5)
# ^ 0011 (3)
# ------
#   0110 (6)
```
