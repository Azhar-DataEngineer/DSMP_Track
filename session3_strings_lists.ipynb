# Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
cur_population = 10000

for year in range(10,0,-1):
  print(year,cur_population)
  cur_population = cur_population/1.1
```

```python
print('Demonstrating break in a nested loop:')
for i in range(1, 4):
  print(f'Outer loop: i = {i}')
  for j in range(1, 4):
    if j == 2:
      print(f'  Inner loop: j = {j}, Breaking inner loop!')
      break  # This breaks only the inner loop (for j)
    print(f'  Inner loop: j = {j}')
  print(f'Outer loop continues after inner loop break (if any).')
print('Loop finished.')
```

```python
# Count the frequency of a particular character in a provided string.
# Eg 'hello how are you' is the string, the frequency of h in this string is 2.


str = input("Enter the text")

char_to_find = input("Enter the character to find")

cnt = 0

for char in str:
  if char == char_to_find:
    cnt = cnt + 1

print(f"the frequency of {char_to_find} is {cnt} in the string")
```

```python
# Write a program which can remove a particular character from a string.

str = input("Enter the text")

char_to_remove = input("Enter the character to remove")

new_str = ''

for char in str:
  if char != char_to_remove:
    new_str = new_str + char

print(new_str)

```
